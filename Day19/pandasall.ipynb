## Filtering Columns and Rows in Pandas

```python
import pandas as pd
df = pd.read_csv('students.csv')
df
```

```python
df[df['Marks'] >= 80]
```

```python
specific_city = ['Mumbai', 'Chennai']

df[df['City'].isin(specific_city)]
```

```python
df[df['City'].str.contains('m')]
```

```python
df = df.set_index('Name')
df
```

```python
df.filter(items=['Name', 'Course', 'Marks'], axis=1)
```

```python
df.filter(like= 'A', axis=0)
```

```python
df = df.set_index('Age')
df
df.loc[22]
```

```python
df.iloc[0]
```

```python
df[df['Marks']  <= 70].sort_values(by=['Course', 'Marks'], ascending= [False, True])
```
