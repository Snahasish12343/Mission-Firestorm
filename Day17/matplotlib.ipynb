## Line Chart

```python
days = np.array([1,2,3,4,5])
Temp = np.array([34,33,27,38,39])

#plt.plot(days, Temp, color= 'Red', ) 
plt.plot(days, Temp, 
         color= "#916ead", 
         linestyle='--',
         linewidth=2,
         marker='o', 
         markersize=8,
         markerfacecolor="#DF2C2C"
         ) 

plt.title("Temperature for Next 5 days", 
          fontsize=20, 
          fontweight="bold", 
          family= 'Arial', 
          color= "red")


plt.xlabel("Days",
           fontsize=20, 
           fontweight='bold',
           family='Arial',
           color="blue")
plt.ylabel("Temperature",
           fontsize=20, 
           fontweight='bold',
           family='Arial',
           color="blue")

plt.show()
```

```python
years = np.array([2021,2022,2023,2024,2025,2026])
maths = np.array([10,12,8,20,22,29])
science = np.array([20,26,26,11,29,25])
english = np.array([23,12,18,29,21,26])


style1 = {
    "marker": "o",
    "markersize": 12,
    "markerfacecolor": 'pink',
    "markeredgecolor": "black",
    "linestyle": "dotted",
    'linewidth': 3
}

plt.plot(years, maths, color='GREEN', **style1, label='Math')
plt.plot(years, science, color='Red', **style1, label='Science')
plt.plot(years, english, color='BLUE', **style1, label='English')
plt.xticks(years)
# plt.yticks()
plt.tick_params(axis='both', colors='Red')  #Use colors
plt.legend()
plt.show()
```

## Bar Chart

```python
#Vertical Bar Chart
```

```python
p = ["laptop", "phone", "tablet", "watch"]
s = [1500, 3200, 800, 2199]


plt.bar(p, s, 
        color=['orange', 'Red', 'Green', 'Blue'], 
        edgecolor='Black', 
        width=0.5)

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Unit Sold")
plt.show()
```

#Horizontal Bar Chart

```python
plt.barh(p, s, 
        color=['orange', 'Red', 'Green', 'Blue'], 
        edgecolor='Black')
plt.title("Product Sales")
plt.ylabel("Product")
plt.xlabel("Unit Sold")
plt.show()
```

## Scatter Plot

```python

height = [150, 160, 165, 170, 175, 180, 185]
weight = [50, 60, 65, 70, 72, 80, 85]
plt.title("Height vs Weidth")
plt.ylabel("Weidth")
plt.xlabel("Height")

plt.scatter(height, weight, color='Red', s=100 , alpha=0.99) #S represents Size & Alpha represents Trasparency
plt.show()
```

```python
aa = [1,2,3,4,5,6,8,10]
ss = [20,30,60,77,78,80,92,98]


ab = [1,1.5,4,5,6,9,11,13]
sb = [30,40,49,52, 56,66,78,84]

plt.scatter(aa, ss, color='Red', alpha=0.5, s=200, label='Class A')

plt.scatter(ab, sb, color='Green', alpha=0.5, s=200, label='Class B')
plt.legend()
plt.show()
```

```python
height = [150, 160, 165, 170, 175, 180, 185]
weight = [50, 60, 65, 70, 72, 80, 85]

plt.scatter(height, weight, c=weight, cmap='plasma', s=200) #S represents Size 
plt.colorbar(label="weight")
plt.show()
```
